```python
import pandas as pd

# Caminho do arquivo
path = "/home/brown/Documents/macro_fred/fred/data/01_raw/fred/DFII10.parquet"

# Carregar
df = pd.read_parquet(path)

print("=== HEAD ===")
display(df.head())

print("\n=== TAIL ===")
display(df.tail())

print("\n=== SCHEMA ===")
df.info()

print("\n=== PERÍODO ===")
print("Início:", df["date"].min())
print("Fim   :", df["date"].max())
print("Total de linhas:", len(df))

print("\n=== VALORES NULOS ===")
print(df.isna().sum())

print("\n=== ESTATÍSTICAS BÁSICAS ===")
display(df["value"].describe())

print("\n=== CHECAGEM DE VALORES INVÁLIDOS ===")
invalid_mask = df["value"].isna()
print("Linhas com value NaN:", invalid_mask.sum())

print("\n=== CONTINUIDADE TEMPORAL ===")
# Diferença entre datas consecutivas
date_diff = df["date"].sort_values().diff().value_counts().head(10)
display(date_diff)

print("\n=== FREQUÊNCIA INFERIDA ===")
print(pd.infer_freq(df["date"].sort_values()))

print("\n=== AMOSTRA DE BURACOS (GAPS > 3 dias) ===")
gaps = df["date"].sort_values().diff()
display(df.loc[gaps > pd.Timedelta(days=3)].head())
```
